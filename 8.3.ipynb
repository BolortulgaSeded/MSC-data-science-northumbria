```python
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_wine
from sklearn.ensemble import AdaBoostClassifier
from sklearn.ensemble import GradientBoostingClassifier 

X, y =load_wine().data,load_wine().target
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)
```

```python
ada_boost = AdaBoostClassifier(random_state=1)
ada_boost.fit(X_train, y_train)
ada_boost.score(X_test, y_test)
```

```python
grad_boost = GradientBoostingClassifier(learning_rate=0.01, random_state=1)
grad_boost.fit(X_train, y_train)
grad_boost.score(X_test, y_test)
```

```python
from xgboost import XGBClassifier
xgb_boost=XGBClassifier(random_state=1,learning_rate=0.01)
xgb_boost.fit(X_train, y_train)
xgb_boost.score(X_test, y_test)
```

```python
from sklearn.model_selection import cross_val_score
from sklearn.ensemble import VotingClassifier

eclf = VotingClassifier(
    estimators=[
        ('Ada Boost', ada_boost),
        ('Grad Boost', grad_boost),
        ('XG Boost', xgb_boost)
    ],
    voting='hard'
)

for clf, label in zip(
    [ada_boost, grad_boost, xgb_boost, eclf],
    ['Ada Boost', 'Grad Boost', 'XG Boost', 'Ensemble']
):
    scores = cross_val_score(clf, X, y, cv=10, scoring='accuracy')
    print("Accuracy: %0.2f (+/- %0.2f) [%s]" % (scores.mean(), scores.std(), label))
```
