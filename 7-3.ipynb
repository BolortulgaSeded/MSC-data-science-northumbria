```python
import pandas as pd
import numpy as np
from sklearn.tree import DecisionTreeRegressor 
from sklearn.model_selection import train_test_split
from sklearn import metrics

mpg = pd.read_csv("auto-mpg.csv")
```

```python
mpg = mpg.drop(columns=['car name'])
mpg['horsepower'] = pd.to_numeric(mpg['horsepower'], errors='coerce')
mpg = mpg.dropna()
```

```python
mpg.head()
```

```python
feature_cols = ['cylinders', 'displacement', 'horsepower', 'weight', 'acceleration', 'model year', 'origin' ]
X=mpg[feature_cols]
y=mpg['mpg']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=3)
```

```python
dt = DecisionTreeRegressor(random_state=3)
dt.fit(X_train, y_train)
y_pred = dt.predict(X_test)
print("Basic Model RMSE:", np.sqrt(metrics.mean_squared_error(y_test, y_pred)))
print("Basic Model R2 Score:", metrics.r2_score(y_test, y_pred))
```

```python
dt = DecisionTreeRegressor(max_depth=8, min_samples_leaf=0.13, random_state=3)
dt.fit(X_train, y_train)
y_pred = dt.predict(X_test)
print("Optimized Model RMSE:", np.sqrt(metrics.mean_squared_error(y_test, y_pred)))
print("Optimized Model R2 Score:", metrics.r2_score(y_test, y_pred))
```
